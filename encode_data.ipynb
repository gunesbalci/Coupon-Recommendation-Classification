```python
import pandas as pd
from dataset_encode import encode_train_test_datasets
from dataset_preprocess import create_train_test_datasets
from dataset_analyze import plot_cramers_v_matrix

df = pd.read_csv("Dataset/in-vehicle-coupon-recommendation.csv")
X_train, X_test, y_train, y_test = create_train_test_datasets(df)
print("Shape of the x_train: " + str(X_train.shape))
print("Shape of the x_test: " + str(X_test.shape))
print("Shape of the y_train: " + str(y_train.shape))
print("Shape of the y_test: " + str(y_test.shape))
```

```python
X_train, X_test = encode_train_test_datasets(X_train, X_test, y_train, True)
print(X_train.head())
for col in X_train.columns:
  unique_values = X_train[col].unique()
  print(f"Sütun: {col}")
  print(f"Unique Değer Sayısı: {len(unique_values)}")
  print(f"Değerler: {unique_values}")
  print("-" * 50)
```

```python
train_df = pd.concat([X_train, y_train], axis=1)
test_df = pd.concat([X_test, y_test], axis=1)
data = pd.concat([train_df, test_df], axis=0).reset_index(drop=True)
plot_cramers_v_matrix(data)
```
